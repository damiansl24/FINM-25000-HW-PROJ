```python
import sys
sys.path.insert(0, '..') # adds the hw1 directory as first position to search.

from src.data_connector_module import load_historical_data as lhd
import matplotlib.pyplot as plt
```

```python
SPY_hist_data = lhd('SPY', 5, '2025-01-01', '2025-03-01')
```

```python
SPY_hist_data.head()
```

```python
df = SPY_hist_data.reset_index()
df.index = df['timestamp']
```

```python
fig, ax1 = plt.subplots(figsize=(12,6))

ax2 = ax1.twinx()

ax1.plot(df['open'], label='Open')
ax1.plot(df['high'], label = 'High')
ax1.plot(df['low'], label = 'Low')
ax1.plot(df['close'], label = 'Close')
ax2.plot(df['volume'], label = 'Volume')


ax1.set_xlabel('Date')
ax1.set_ylabel('Price')
ax2.set_ylabel('Volume')
ax1.set_title('SPY OHLCV data, between 2025-01-01 and 2025-03-01')
ax1.legend()

plt.show()
```
